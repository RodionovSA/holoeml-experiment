```python
import pco
import numpy as np
import matplotlib.pyplot as plt
import time

from instruments.pypcocam import PcoCamera

SERIAL = "22500190"
```

```python
camera = PcoCamera(serial=SERIAL)
camera.set_roi(1000, 1000)
```

```python
delay = 0.5
max_exp = 1000


camera.set_exposure_ms(200)
camera.arm()
for i in range(4):
    camera.set_exposure_ms(200)
    time.sleep(delay)     
    image_short = camera.get_image()
    camera.set_exposure_ms(max_exp)
    time.sleep(delay) 
    image_long = camera.get_image()
    print(f"Mean short: {np.mean(image_short[...,0])}, Std short: {np.std(image_short[...,0])}")
    print(f"Mean long: {np.mean(image_long[...,0])}, Std long: {np.std(image_long[...,0])}\n")
camera.disarm()
```

```python
camera.close()
```
