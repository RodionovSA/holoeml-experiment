```python
import pco
import matplotlib.pyplot as plt

SERIAL = "22500190"
```

```python
camera = pco.Camera(serial=int(SERIAL))
camera.auto_exposure_off()
```

```python
camera.exposure_time = 0.2
```

```python
camera.configuration = {"roi": (1, 1, 4416, 2368)}
```

```python
camera.
```

```python
camera.configuration
```

```python
camera.record(mode="sequence", number_of_images=5)
im, meta = camera.images(crop=[500, 1, 3800, 2368])
```

```python
plt.imshow(im[0], cmap='gray')
plt.colorbar()
```

```python
camera.close()
```
