```python
import time
from instruments.config.config import load_equipment
from instruments.shutter import ShutterControl
```

```python
sh = ShutterControl(load_equipment().shutter_port)   # COM3

# --- info / state ---
print(sh.identify())
print(sh.get_mode(), sh.is_unlocked(), sh.is_open(1), sh.is_open(2))
```

```python
sh.set_mode('manual')                       # closes both shutters while switching
sh.open_all()
sh.shut_all()
sh.open_all()
```

```python
# --- internal mode: free-run at 10 Hz ---
sh.lock()
sh.set_mode('internal')                     # closes both shutters while switching
sh.unlock()
print(sh.configure_internal(1, freq_hz=10, duty=50, cycles=0))   # cycles=0 -> runs until stop
sh.start(1)
time.sleep(3)
sh.stop(1)
print('cycles done:', sh.get_completed_cycles(1))
```
