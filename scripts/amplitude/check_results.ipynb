```python
import time
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load(r"D:\sergei\holoeml-experiment\sample_xpol_20260802_170155.npz")
img = data['images'][...,0]
```

```python
# Plot at a few wavelenghts
wavs = data['wavelengths']
square_roi_size = 100
square_roi_center = (1550, 1075)
roi = (square_roi_center[0] - square_roi_size//2, square_roi_center[0] + square_roi_size//2,
       square_roi_center[1] - square_roi_size//2, square_roi_center[1] + square_roi_size//2)
bands = [0, 10, 50, 80, -1]
fig, ax = plt.subplots(1,5,figsize=(16, 6))
for i, b in enumerate(bands):
    ax[i].imshow(img[b, roi[0]:roi[1], roi[2]:roi[3]], cmap='gray')
    ax[i].set_title(f'wavelength {wavs[b]:.1f} nm\n band {b}')

# plt.imshow(img[10, roi[0]:roi[1], roi[2]:roi[3]], cmap='gray')
# plt.colorbar()
# plt.xlim([1500, 2500])
# plt.ylim([1000, 2000])
```
