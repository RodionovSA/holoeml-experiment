```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.load('phase_measurement_20260805_121340.npz') 
phi = data['phi']
```

```python
data2 = np.load('phase_measurement_20260805_122048.npz') 
phi2 = data2['phi']
```

```python
plt.imshow(phi, cmap='gray')
```

```python
y_coord = 850
x_coord_0 = 350
x_coord_1 = 550
delta_x = 20
delta_y = 40

plt.figure(figsize=(12, 6))
plt.subplot(121)
plt.imshow(phi, cmap='gray')
plt.hlines(y_coord, x_coord_0, x_coord_1, colors='r', linestyles='dashed')
plt.xlim(x_coord_0 - delta_x, x_coord_1 + delta_x)
plt.ylim(y_coord + delta_y, y_coord - delta_y)

plt.subplot(122)
plt.plot(np.degrees(phi[y_coord, x_coord_0:x_coord_1]), label='phi')
plt.plot(np.degrees(phi2[y_coord, x_coord_0:x_coord_1]), label='phi')

plt.tight_layout()
```
